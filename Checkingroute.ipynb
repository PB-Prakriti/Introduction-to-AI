```python
# Road network represented using an adjacency list

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C'],
    'E': ['F'],
    'F': ['E'],
    'G': []
}


# 1. DFS to check whether a route exists
def route_exists(graph, start, destination):
    visited = set()

    def dfs(city):
        # If destination is reached
        if city == destination:
            return True

        visited.add(city)

        # Visit all connected cities
        for neighbour in graph[city]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(start)


# 2. Find all connected components
def find_connected_components(graph):
    visited = set()
    components = []

    for city in graph:
        if city not in visited:

            component = []

            def dfs(city):
                visited.add(city)
                component.append(city)

                for neighbour in graph[city]:
                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)
            components.append(component)

    return components


# Take input from the user
start = input("Enter starting city: ").upper()
destination = input("Enter destination city: ").upper()

# Check if the cities exist
if start not in graph or destination not in graph:
    print("Invalid city name.")

else:
    # Check route
    if route_exists(graph, start, destination):
        print(f"Route exists from {start} to {destination}.")
    else:
        print(f"No route exists from {start} to {destination}.")


# Find connected components
components = find_connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, start=1):
    print(f"Component {i}: {component}")
```
